```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
#1
x = np.array([1,2,3,4])
y = np.array([10,20,25,30])
plt.plot(x,y)
plt.show()
```

```python
#2
x = np.array([1,2,3,4])
y = np.array([10,20,25,30])
plt.plot(x,y)
plt.title("Simple Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#3
x = np.array([0,1,2,3])
y = np.array([3,6,1,8])
plt.plot(x,y,'r:')
plt.title("Red dotted Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#4 and #5
x = np.array([1,2,3,4])
y1 = np.array([10,20,25,30])
y2 = np.array([5,15,20,25])
plt.plot(x,y1,'r-',label='Line 1')
plt.plot(x,y2,'c-',label='Line 2')
plt.title("Two Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
#15
x=np.array([1,2,3,4])
y=np.array([1,4,9,16])
plt.figure(figsize=(5, 5))  # Set width=5 inches, height=5 inches
plt.plot(x, y,color='darkred', linestyle=':',marker='o',linewidth=3,markersize=6)
plt.title("Simple Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
#16
x=np.array([0,1,2,3,4,5,6])
y=np.array([0,0.001,0.1,1,0.1,0.001,0])
plt.figure(figsize=(6, 6))  # Set width=5 inches, height=5 inches
plt.plot(x, y,color='darkred', linestyle=':',marker='o',linewidth=3,markersize=6)
plt.title("Simple Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
indexmax=np.argmax(y)
xmax=x[indexmax]
ymax=y[indexmax]
ann= "Peak point: ("+str(xmax)+","+str(ymax)+")"
plt.annotate(ann,
             xy=(xmax,ymax),
             xytext=(xmax+0.1,ymax+0.1),
             arrowprops=dict(facecolor='black', arrowstyle='->'),
             fontsize=10,
             color='red')
plt.tight_layout()
plt.show()
```

```python
#17
Fruits = np.array(['Apple', 'Banana', 'Mango'])
Quantity = np.array([10, 20, 15])
plt.figure(figsize=(5, 5))
bars = plt.bar(Fruits, Quantity, color='skyblue',width=0.5)
for i in range(len(Quantity)):
    plt.text(i, Quantity[i] + 1, str(Quantity[i]), ha='center', va='bottom', fontsize=10)

plt.title("Bar Chart with Value Labels")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.ylim(0,max(Quantity)+5)
plt.tight_layout()
plt.show()
```

```python
#18
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [250, 300, 150, 400]
plt.figure(figsize=(5, 5))
plt.bar(months, sales, color='skyblue',width=0.5)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
for i in range(len(sales)):
    plt.text(i, sales[i] + 10, str(sales[i]), ha='center', fontsize=9)
plt.tight_layout()
plt.show()
```

```python
#19
Fruits = np.array(['Apple', 'Banana', 'Mango'])
Quantity = np.array([10, 20, 15])
plt.figure(figsize=(15, 5))
plt.subplot(1,3,1)
plt.plot(Fruits,Quantity,'r:',linewidth = 3, marker='o',markersize=6)
plt.title("Fruits vs Quantity")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.subplot(1,3,2)
plt.bar(Fruits,Quantity,color='g',width=0.5)
plt.title("Fruits vs Quantity")
plt.xlabel("Fruits")
plt.ylabel("Quantity")
plt.subplot(1,3,3)
plt.pie(Quantity,labels=Fruits,colors=['orange', 'green', 'pink'],shadow=True)
plt.title("Fruits Pie chart")
plt.tight_layout()
plt.show()
```
