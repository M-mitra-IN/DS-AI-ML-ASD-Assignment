```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#1
df = sns.load_dataset('tips')
print(df.head(5))
```

```python
#2
sns.scatterplot(x='total_bill',y='tip',data=df)
plt.title("Scatterplot of Tips vs Total Bill")
plt.show()
```

```python
#3
sns.lineplot(x='total_bill',y='tip',data=df)
plt.title("Lineplot of Tip vs Total")
plt.show()
```

```python
#4
sns.barplot(x='day',y='tip',data=df)
plt.title("Barplot of Tip vs Day")
plt.show()
```

```python
#5
sns.countplot(x='day', data = df)
plt.title("Number of records availabe for each day")
plt.show()
```

```python
#6
sns.boxplot(x='day',y='total_bill',data=df)
plt.title("Total Bill distribution across different days")
plt.show()
```

```python
#7
sns.violinplot(x='day',y='total_bill',data=df)
plt.title("Total Bill vs Days")
plt.show()
```

```python
#8
sns.histplot(x='total_bill',data=df,bins = 10 ,kde= True)
plt.title("Histogram")
plt.show()
```

```python
#9
df = sns.load_dataset('iris')
sns.pairplot(df,hue='species')
plt.title("Pair Plot of all features in iris dataset")
plt.show()
```

```python
#10
df= sns.load_dataset('titanic')
corr = df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title("Correlation Matrix of The titanic incident")
plt.show()
```

```python
#11
df = sns.load_dataset('penguins')
sns.scatterplot(x='flipper_length_mm',y='bill_length_mm',data=df)
plt.title("Scatterplot of Bill Length vs Flipper Length of Penguins")
plt.show()
```

```python
#12
df= sns.load_dataset('titanic')
sns.boxplot(x='class',y='fare',data=df)
plt.title("Boxplot of Fare vs Class of Titanic")
plt.show()
```

```python
#13
df = sns.load_dataset('titanic')
sns.countplot(x='survived',data=df)
plt.title("Countplot of Survivors and Non-survivors of Titanic Incident")
plt.show()
```

```python
#14
df = sns.load_dataset('tips')
g = sns.FacetGrid(df, col='sex')
g.map(plt.hist, 'total_bill', bins=10, color='skyblue', edgecolor='black')
plt.suptitle("Histogram of Total Bill by Gender", fontsize=16, y=1.05)
plt.show()
```

```python
#15
df= sns.load_dataset('penguins')
corr = df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title("Penguins Correlation Matrix")
plt.show()
```

```python
#16
df= sns.load_dataset('diamonds')
sns.violinplot(x='cut',y='price',data=df)
plt.title("Diamond Price by Cut")
plt.show()
```

```python
#17
df = sns.load_dataset('penguins')
sns.pairplot(df,hue='species')
plt.title("Pair Plot of all features of penguins")
plt.show()
```

```python
#18
df = sns.load_dataset("flights")
pivot_table = df.pivot(index="month", columns="year", values="passengers")
plt.figure(figsize=(12, 6))
sns.heatmap(pivot_table, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Heatmap of Monthly Flight Passengers Over the Years")
plt.show()
```

```python
#19
sns.set_style('dark')
df = sns.load_dataset('penguins')
sns.scatterplot(x='flipper_length_mm',y='bill_length_mm',data=df)
plt.title("Scatterplot of Bill Length vs Flipper Length of Penguins")
plt.show()
```

```python
#20
import pandas as pd
df = sns.load_dataset("tips")
df_numeric = pd.get_dummies(df, drop_first=True)
corr_matrix = df_numeric.corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Full Correlation Heatmap of Tips Dataset (Numeric)")
plt.show()
```
